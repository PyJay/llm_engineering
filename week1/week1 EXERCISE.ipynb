```python
# imports
from openai import OpenAI
from IPython.display import display, Markdown, update_display
```

```python
# constants
MODEL_GPT = 'gpt-4o-mini'
MODEL_LLAMA = 'llama3.2'
```

```python
# set up environment
OLLAMA_BASE_URL = "http://localhost:11434/v1"
openai = OpenAI()
ollama = OpenAI(base_url=OLLAMA_BASE_URL)
system_prompt = """
You are a professional yet very helpful highly technical and experienced software engineer who is an expert in AI and LLMs.
You have a great ability to explain complex technical concepts in a clear concise and easy to follow manner.
You exclusively cater to software engineers who are looking to upskill and learn about the latest advancements in AI and LLMs.
You always respond in markdown format with code blocks where appropriate.
"""
system_message = {"role": "system", "content": system_prompt}

def stream_from(prompt, provider, model):
    stream = provider.chat.completions.create(
        model=model,
        messages=[
            system_message,
            {"role": "user", "content": prompt}
        ],
        stream=True
    )
    response = ""
    display_handle = display(Markdown(""), display_id=True)
    for chunk in stream:
        response += chunk.choices[0].delta.content or ''
        update_display(Markdown(response), display_id=display_handle.display_id)

```

```python
# here is the question; type over this to ask something new

question = """
Please explain what this code does and why:
yield from {book.get("author") for book in books if book.get("author")}
"""
```

```python
# Get gpt-4o-mini to answer, with streaming

answer = stream_from(question, provider=openai, model=MODEL_GPT)
```
